```python
# Análisis de clientes
clientes = (
    df.groupby('CustomerID').agg(
        ingresos_totales=('TotalVenta', 'sum'),
        unidades_compradas=('Quantity', 'sum'),
        numero_pedidos=('InvoiceNo', 'nunique'),
        productos_diferentes=('Description', 'nunique')
    )
    .sort_values('ingresos_totales', ascending=False)
)

clientes.head(10)
```

```python
frecuencia_clientes = (
    clientes['numero_pedidos']
    .value_counts()
    .sort_index()
)

frecuencia_clientes
```

```python
clientes_una_compra = (clientes['numero_pedidos'] == 1).sum()
clientes_recurrentes = (clientes['numero_pedidos'] > 1).sum()

print("Clientes con una sola compra:", clientes_una_compra)
print("Clientes recurrentes:", clientes_recurrentes)
print("Total de clientes:", len(clientes))
```

```python
# Ordenar clientes de mayor a menor facturación
clientes_ordenados = clientes.sort_values(
    'ingresos_totales',
    ascending=False
).copy()

# Ingresos acumulados
clientes_ordenados['ingresos_acumulados'] = (
    clientes_ordenados['ingresos_totales'].cumsum()
)

# Porcentaje acumulado de ingresos
clientes_ordenados['%_ingresos_acumulados'] = (
    clientes_ordenados['ingresos_acumulados']
    / clientes_ordenados['ingresos_totales'].sum()
    * 100
)

# Cantidad de clientes
total_clientes = len(clientes_ordenados)

# Resultados
for porcentaje in [10, 20, 50]:
    cantidad = int(total_clientes * porcentaje / 100)

    ingreso = clientes_ordenados.iloc[cantidad - 1]['%_ingresos_acumulados']

    print(
        f"El {porcentaje}% de los clientes genera "
        f"el {ingreso:.2f}% de los ingresos."
    )
```

```python
# Fecha de referencia: día siguiente a la última compra registrada
fecha_referencia = df_ventas['InvoiceDate'].max() + pd.Timedelta(days=1)

# Calcular RFM
rfm = (
    df_ventas[df_ventas['Quantity'] > 0]
    .groupby('CustomerID')
    .agg(
        Recency=('InvoiceDate', lambda x: (fecha_referencia - x.max()).days),
        Frequency=('InvoiceNo', 'nunique'),
        Monetary=('TotalVenta', 'sum')
    )
)

rfm.head()
```

```python
rfm.describe()
```

```python
# Crear puntuaciones RFM de 1 a 5

rfm['R_score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1],
    duplicates='drop'
)

rfm['F_score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5],
    duplicates='drop'
)

rfm.head()
```

```python
rfm['RFM_Score'] = (
    rfm['R_score'].astype(str) +
    rfm['F_score'].astype(str) +
    rfm['M_score'].astype(str)
)

rfm.head()
```

```python
def segmentar_cliente(row):
    r = int(row['R_score'])
    f = int(row['F_score'])
    m = int(row['M_score'])

    # Clientes de máximo valor
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'

    # Clientes frecuentes y valiosos
    elif r >= 3 and f >= 4 and m >= 3:
        return 'Clientes Leales'

    # Clientes recientes con potencial
    elif r >= 4 and f <= 3 and m >= 3:
        return 'Potenciales'

    # Clientes que antes tenían valor pero están perdiendo actividad
    elif r <= 2 and f >= 3 and m >= 3:
        return 'En Riesgo'

    # Clientes con baja actividad general
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Perdidos'

    # Resto
    else:
        return 'Clientes Ocasionales'


rfm['Segmento'] = rfm.apply(segmentar_cliente, axis=1)

rfm.head(10)
```

```python
segmentos = (
    rfm['Segmento']
    .value_counts()
    .to_frame('Clientes')
)

segmentos['Porcentaje'] = (
    segmentos['Clientes'] / len(rfm) * 100
).round(2)

segmentos
```

```python
valor_segmentos = (
    rfm.groupby('Segmento')
    .agg(
        Clientes=('Monetary', 'count'),
        Ingresos=('Monetary', 'sum'),
        Frecuencia_promedio=('Frequency', 'mean'),
        Recency_promedio=('Recency', 'mean')
    )
    .sort_values('Ingresos', ascending=False)
)

valor_segmentos['%_Ingresos'] = (
    valor_segmentos['Ingresos']
    / valor_segmentos['Ingresos'].sum()
    * 100
).round(2)

valor_segmentos
```

```python
import matplotlib.pyplot as plt

# Ordenar segmentos de mayor a menor ingreso
grafica_ingresos = valor_segmentos.sort_values(
    'Ingresos',
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    grafica_ingresos.index,
    grafica_ingresos['Ingresos']
)

plt.title('Ingresos totales por segmento RFM')
plt.xlabel('Segmento de clientes')
plt.ylabel('Ingresos (£)')
plt.xticks(rotation=45)

# Mostrar valores encima de las barras
for i, valor in enumerate(grafica_ingresos['Ingresos']):
    plt.text(
        i,
        valor,
        f'£{valor:,.0f}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Ordenar de mayor a menor porcentaje de ingresos
grafica_porcentaje = valor_segmentos.sort_values(
    '%_Ingresos',
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    grafica_porcentaje.index,
    grafica_porcentaje['%_Ingresos']
)

plt.title('Distribución porcentual de los ingresos por segmento RFM')
plt.xlabel('Segmento de clientes')
plt.ylabel('Porcentaje de ingresos (%)')
plt.xticks(rotation=45)

# Mostrar porcentaje encima de cada barra
for i, valor in enumerate(grafica_porcentaje['%_Ingresos']):
    plt.text(
        i,
        valor,
        f'{valor:.2f}%',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for segmento in rfm['Segmento'].unique():
    datos = rfm[rfm['Segmento'] == segmento]

    plt.scatter(
        datos['Recency'],
        datos['Monetary'],
        label=segmento,
        alpha=0.6
    )

plt.title('Recency vs. Monetary por segmento RFM')
plt.xlabel('Recency (días desde la última compra)')
plt.ylabel('Monetary (£)')
plt.legend()
plt.tight_layout()
plt.show()
```

## Conclusiones y recomendaciones generales

###Conclusiones

- El análisis de clientes permitió identificar una alta concentración de los ingresos en un grupo reducido de compradores. El 10% de los clientes genera aproximadamente el 60,10% de los ingresos, mientras que el 20% concentra cerca del 73,90%, lo que evidencia la importancia de implementar estrategias específicas de fidelización para los clientes de mayor valor.

- La segmentación RFM permitió identificar que los Champions representan 957 clientes y generan aproximadamente el 65,17% de los ingresos, convirtiéndose en el segmento de mayor importancia para el negocio. Estos clientes presentan compras recientes, una frecuencia elevada y un alto valor monetario.

- Por otra parte, se identificaron 453 clientes En Riesgo, que representan aproximadamente el 8,32% de los ingresos. Su recency promedio de 141 días indica que han pasado períodos importantes sin realizar compras, por lo que constituyen una oportunidad para desarrollar estrategias de recuperación antes de que pasen al segmento de clientes perdidos.

- Los 277 clientes Potenciales presentan una recency promedio de apenas 17 días, pero una frecuencia relativamente baja. Esto indica que son clientes que han comprado recientemente y que podrían aumentar su valor mediante estrategias orientadas a incrementar la frecuencia de compra.

- Finalmente, los 821 clientes Perdidos generan solamente el 2,11% de los ingresos, por lo que representan una prioridad menor frente a los clientes Champions, En Riesgo y Potenciales.

###Recomendaciones

1. Fidelizar a los clientes Champions

Implementar programas de fidelización, beneficios exclusivos, promociones personalizadas y acceso anticipado a nuevos productos para conservar a los clientes que actualmente generan la mayor proporción de los ingresos.

2. Recuperar clientes En Riesgo

Diseñar campañas de reactivación mediante descuentos personalizados, recomendaciones de productos y comunicaciones dirigidas. El objetivo debe ser reducir el tiempo desde la última compra y evitar que estos clientes pasen al segmento Perdidos.

3. Convertir clientes Potenciales en clientes Leales

Estimular una segunda y tercera compra mediante promociones relacionadas con productos adquiridos anteriormente, recomendaciones personalizadas y estrategias de venta cruzada.

4. Aumentar el valor de los clientes Ocasionales

Este es el segmento más numeroso, con 1.377 clientes, pero representa solamente el 9,30% de los ingresos. Se recomienda trabajar en estrategias de frecuencia de compra para convertir una mayor proporción de estos clientes en clientes recurrentes.

5. Priorizar los recursos de marketing

La concentración de ingresos demuestra que no todos los clientes tienen el mismo valor para el negocio. Por ello, las acciones comerciales deberían priorizar la retención de clientes de alto valor y la recuperación de clientes con potencial, en lugar de distribuir los recursos de manera uniforme.

###Conclusión general

El análisis demuestra que la gestión de clientes debe ir más allá de observar únicamente las ventas totales. La combinación del análisis de comportamiento de compra y la segmentación RFM permitió identificar qué clientes generan mayor valor, cuáles presentan riesgo de abandono y cuáles tienen potencial de crecimiento. Estos resultados proporcionan una base para desarrollar estrategias comerciales diferenciadas, mejorar la fidelización y orientar los recursos hacia los segmentos con mayor impacto sobre los ingresos.
